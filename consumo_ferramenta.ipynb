# Análise Exploratória de Dados (EDA)

```python
# importa biblioteca necessária para EDA
import pandas as pd
```

```python
# importa os dados
df_producao = pd.read_parquet('tb_producao_consolidado.parquet') # importa os dados de produção
df_consumo_ferramentas = pd.read_parquet('tb_consumo_ferramentas_consolidado.parquet') # importa os dados de consumo de ferramentas

print('--- Informações: df_producao ---' + '\n') # exibe um descritivo do df
df_producao.info() # exibe o .info() do df

print('\n' + '-'*40 + '\n') # exibe um separador

print('--- Informações: df_consumo_ferramentas ---' + '\n') # exibe um descritivo do df
df_consumo_ferramentas.info() # exibe o .info() do df
```

Não será necessário realiza limpeza de dados NA por conta do script 'consolidar_arquivos.py' já realiza essa operação durante a consolidação dos dados.

```python
# converte as conlunas de data para tipo datetime
df_producao['Dt.lçto.'] = pd.to_datetime(df_producao['Dt.lçto.'], format='%Y-%m-%d')
df_consumo_ferramentas['Data de lançamento'] = pd.to_datetime(df_consumo_ferramentas['Data de lançamento'], format='%Y-%m-%d')
```
